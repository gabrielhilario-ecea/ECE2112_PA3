```python
import pandas as pd

cars = pd.read_csv('cars.csv')
```

## A. Positional and Label-Based Slicing

```python
# A(a): shape and complete list of column names
print('Shape:', cars.shape)
print('Columns:', cars.columns.tolist())
```

```python
# A(b): rows 6 through 10, where the first data row is row 1
cars_6_to_10 = cars.iloc[5:10]
cars_6_to_10
```

```python
# A(c): requested columns, selected by labels
cars_6_to_10[['Model', 'mpg', 'cyl', 'hp', 'gear']]
```

## B. Model Lookup

```python
# B(a): complete Toyota Corolla row, found through Boolean indexing
toyota = cars[cars['Model'] == 'Toyota Corolla']
toyota
```

```python
# B(b): requested Pontiac Firebird columns, found through Boolean indexing
pontiac = cars.loc[cars['Model'] == 'Pontiac Firebird', ['Model', 'mpg', 'hp', 'wt']]
pontiac
```

## C. Multi-Model Subsetting

```python
# Select the three requested models by their values and retain the required columns
models = ['Datsun 710', 'Lotus Europa', 'Ferrari Dino']
selected_cars = cars.loc[cars['Model'].isin(models), ['Model', 'mpg', 'cyl', 'hp', 'gear']]

display(selected_cars)
print('Shape:', selected_cars.shape)
assert selected_cars.shape == (3, 5), 'selected_cars must contain exactly 3 rows and 5 columns.'
```
